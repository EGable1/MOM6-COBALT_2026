```python


def mom6_daily_to_df(file_path):
    ds = xr.open_dataset(file_path)

    # extract variable
    da = ds["btm_o2"]

    # convert to dataframe (keeps y/x grid)
    dfm = da.to_dataframe().reset_index()

    #drop na values of oxygen and convert from g/L to mg/L
    dfm=df.dropna(subset=['btm_o2'])
    dfm['btm_o2']=df['btm_o2']*1000

    # extract date from filename
    date_str = file_path.split("HINDCAST-")[-1].replace(".nc", "")
    dfm["time"] = pd.to_datetime(date_str)

    # reorder cleanly
    return dfm[["time", "lat", "lon", "btm_o2"]]
```
